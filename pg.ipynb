```python
samples_names = ['hh_1', 'high_1']
run_output = Path("/home/hoshi/ME/Master/runs/test_run_e100")
```

```python
for sample in samples_names:
  simulated = 'K' not in sample
  test_models(
    data_path=f'data/sim2/{sample}',
    hypers=hypers,
    output_dir="/home/hoshi/ME/Master/runs/test_run_e100",
    models_versions=models_versions,
    simulated=simulated
    )
```

```python
for sample in samples_names:
  simulated = 'K' not in sample
  data = prep_data(f'data/sim2/{sample}', simulated=simulated)
  for model in models_versions:
    
    evaluate_model(
        model= run_output / f"models/{sample}_{model['label']}",
        sample=sample,
        data = data,
        data_path = Path(f"data/sim2/{sample}"),
        model_label=model['label'],
        input_dir= run_output,
        simulated=simulated
        )
```
